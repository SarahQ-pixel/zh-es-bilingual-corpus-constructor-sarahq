### Google Drive Loading (Optional):

Example: create a symbolic link from a Google Drive folder to a project folder. 
 Replace <your_project_path> and <your_google_drive_path> with your actual paths

```python
from google.colab import drive

drive.mount('/content/drive')

import os
import shutil

project_target_path = "<your_project_path>"

google_drive_source_path = "<your_google_drive_path>"

if os.path.exists(project_target_path):
    shutil.rmtree(project_target_path)

os.symlink(
    google_drive_source_path,
    project_target_path,
    target_is_directory=True
)

print(
    f"{project_target_path} -> {google_drive_source_path}"
)
```
